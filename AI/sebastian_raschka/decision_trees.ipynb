## Recursive Algorithms & Big-O

Recursive algorithms are related to the concept called divide and conquer. Meaning, dividing a big problem into smaller sub problems and solving one problem at a time. It can be implemented by using Recursion.

We should not use recursion all time, because python has limits on recursion. so if we pass a large list, for each iteration a new stack is created on top. once created, the execution starts from top to bottom. if too many stack is created, it will raise a stackoverflow error.

````python
# Example Recursion function
def some_fun(x):
    if x == 0:
        return 0
    else:
        return x + some_fun(x - 1)


print(some_fun(5))  # Output: 15
````

````python
def len_array(arr):
    if arr == []:
        return 0
    else:
        return 1 + len_array(arr[1:])


print(len_array([1, 2, 3, 4, 5]))  # Output: 5
````

````python
# Divide and Conquer algorithm (Quick Sort):
def quick_sort(arr):
    if len(arr) < 2:
        return arr
    else:
        pivot = arr[0]
        smaller, bigger = [], []
        for i in arr[1:]:
            if i <= pivot:
                smaller.append(i)
            else:
                bigger.append(i)
        return quick_sort(smaller) + [pivot] + quick_sort(bigger)


print(quick_sort([3, 6, 8, 10, 1, 2, 1]))  # Output: [1, 1, 2, 3, 6, 8, 10]
````



- X = List of features (ex: color, size, shape), y = List of targets (apple, orange)
- The recursive function first check if all the elements in the node is same, if same then returns the tree
- If the features in the nodes are different, then it selects the best feature in X that splits y by best.
- It splits the X into two groups left and right, left if NO and right if YES.
- send both groups again to the recursive function
